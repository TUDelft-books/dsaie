## Effective Prompting
When interacting with a Large Language Model (LLM) about your code, the way you prompt the model can significantly impact the quality of its responses. We recommend using the integrated AI chat panel (available in Cursor or other editors) instead of external tools like ChatGPT. This allows the LLM to directly reference your project files and the broader code context.

If you’re using the Cursor editor, you can access the AI Chat panel in the top-right corner by toggling the AI Panel button (next to the gear icon). In the bottom left corner of the chat box, ensure you’re using the latest GPT model. From there, you can:
* Ask questions (or follow-ups) about the code
* Reference files with the '+ Add Context' button
* Create a new chat to reset the context of the current chat in the top right '+' button.

Try asking the LLM:
> "How do you calculate the mean of a tensor?"

This leads us to the first key concept in effective prompting: **specifying context**. In the example above, we did not specify which library we were using, so the LLM might return a solution using NumPy or TensorFlow instead of PyTorch. We can refine this by adding more context:
> "How do you calculate the mean of a tensor in PyTorch?"

Being specific with your prompts typically results in better responses. You can also take your prompts a step further by specifying the desired **output format**:
> "How do you calculate the mean of a tensor in PyTorch? Return two examples with comments on every line."

**All three options (Cursor AI, Copilot, and Gemini) come preconfigured with roles, so you can skip this the role**. However, when using a more general chat interface like ChatGPT, it helps to explicitly instruct the LLM to take on this role:
> You are an AI coding assistant specializing in PyTorch. 
>
> How do you calculate the cross entropy loss in PyTorch?
>
> Return a code block with no additional explanation in your response other than comments on every line."


More generally, one possible effective format you can stick to when prompting an LLM could be,
1. **Role**: Define the role of the AI (e.g., coding assistant). *Skip this for now.*
2. **Task**: State the problem or question.
3. **Format/ decorations**: Specify the desired output format or any additional instructions (e.g., code with comments, brief explanation, etc.).

In this workshop, we’ll assume you are using the LLM features directly within the coding editor, where the role and format are often handled automatically in the background. Therefore, you can focus on providing a clear task, and optionally, specify the output format for more tailored responses.

## The Importance of Human-in-the-Loop

While AI coding assistants can significantly speed up development, it is important to remember that these tools are meant to assist, not replace, human programmers. AI models, including large language models (LLMs), can make critical errors—often in unexpected ways. To maintain high-quality code, it’s essential to always keep a **human-in-the-loop**, reviewing, testing, and validating any AI-generated code.

This means you should double-check the correctness of the code produced by LLMs, ensuring that it functions as intended and aligns with best practices. AI coding assistants can be powerful, but they are far from infallible.
