# Lecture 1. Decision Tree
A decision tree is a non-parametric, non-linear algorithm used in supervised machine learning, which can be used for both classification and regression tasks. It constructs a hierarchical tree structure, where each internal node represents a test on an attribute. Branches from these nodes indicate the outcomes of the tests, and each leaf node, or terminal node, contains a class label or a value, depending on the task.

In this lecture, we'll learn the concept of decision trees, delve into classification trees and regression trees, and explore how to construct them. Additionally, we'll discuss pruning techniques, which are crucial for preventing overfitting, to ensure our models remain robust and effective.

The outline of this short course is defined as follows:

- **Section 1**: Introduction to decision tree
- **Section 2**: Classification tree
- **Section 3**: Regression tree
- **Section 4**: Combat overfitting
