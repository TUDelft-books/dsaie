# Lecture 2. Ensemble methods

In this lecture, we'll introduce **Ensemble Methods**. We begin by introducing the concept of **ensemble learning**, covering the foundational concepts, and explaining how combining multiple models can lead to more accurate and reliable predictions than any single model alone. Next, we dive into **Bagging**, a technique that improves the stability and accuracy of machine learning algorithms by creating multiple subsets of data and aggregating their predictions. Building on the principles of Bagging, we explore **Random Forests**, a powerful bagging-based method known for its simplicity and effectiveness across various applications. We then shift our focus to another type of ensemble method - **Boosting**, which sequentially builds models to correct the errors of prior ones, thereby enhancing prediction accuracy. We'll also study a specific boosting-based method - **Gradient Boosted Tree**, which optimizes the prediction performance of prior trees by minimizing loss functions. Finally, we conclude with a brief introduction to **model complexity** and **feature importance**.

The outline of this short course is defined as follows:

- **Section 1**: Ensemble learning
- **Section 2**: Bagging
- **Section 3**: Random forest
- **Section 4**: Boosting
- **Section 5**: Gradient boosted tree
- **Section 6**: Model complexity
- **Section 7**: Feature importance
